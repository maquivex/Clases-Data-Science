```python
import joblib
import numpy as np

# Load the best model and scalers
model = joblib.load('model.pkl')
scaler_X = joblib.load('scaler_X.pkl')
scaler_y = joblib.load('scaler_y.pkl')

def predict_price(new_data):
  # Ensure new_data is a 2D array for scaling
  new_data_array = np.array(new_data).reshape(1, -1)

  # Scale the input data
  new_data_scaled = scaler_X.transform(new_data_array)

  # Make prediction using the model
  prediction_scaled = model.predict(new_data_scaled)

  # Inverse transform the prediction to get original price scale
  prediction = scaler_y.inverse_transform(prediction_scaled.reshape(1, -1))

  return prediction[0][0]
```

```python
import gradio as gr
import numpy as np

def gradio_predict(
    sistema,
    procesador,
    marca,
    banda,
    nucleos,
    memoria_ram,
    memoria_interna,
    bateria_tamaño,
    camera_trasera,
    camera_frontal,
    tamano_pantalla
):
    # Construir la lista de datos en el formato esperado por el modelo
    data_list = os_name[sistema] + \
                processor_name[procesador] + \
                brand_name[marca] + \
                [banda, nucleos, memoria_ram, memoria_interna, bateria_tamaño, camera_trasera, camera_frontal, tamano_pantalla]

    # Convertir a array de numpy
    new_data_array = np.array(data_list)

    # Realizar la predicción y ajustar la escala
    predicted_price = predict_price(new_data_array) * 10000

    return f"Precio estimado: {predicted_price:.3f} EUROS"

# Definir los componentes de entrada para Gradio
inputs = [
    gr.Dropdown(list(os_name.keys()), label="Sistema Operativo"),
    gr.Dropdown(list(processor_name.keys()), label="Procesador"),
    gr.Dropdown(list(brand_name.keys()), label="Marca"),
    gr.Number(label="Banda (1 para 4G/5G, 0 para 3G/2G)", value=1),
    gr.Number(label="Núcleos del Procesador", value=4),
    gr.Number(label="Memoria RAM (GB)", value=12),
    gr.Number(label="Memoria Interna (GB)", value=128),
    gr.Number(label="Tamaño Batería (mAh)", value=5000),
    gr.Number(label="Cámara Trasera (MP)", value=50),
    gr.Number(label="Cámara Frontal (MP)", value=20),
    gr.Number(label="Tamaño Pantalla (pulgadas)", value=6.80)
]

# Definir el componente de salida
output = gr.Textbox(label="Precio Predicho")

# Crear la interfaz de Gradio
interface = gr.Interface(
    fn=gradio_predict,
    inputs=inputs,
    outputs=output,
    title="Estimador de Precios de Smartphones",
    description="Introduce las características de un smartphone para estimar su precio."
)

# Lanzar la interfaz
interface.launch()
```
